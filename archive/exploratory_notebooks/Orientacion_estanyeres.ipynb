```python
import numpy as np
import pandas as pd

# Orientaciones dominantes (modo) de Estanyeres A-D
orientaciones = {
    'Estanyeres_A': 132.37,
    'Estanyeres_B': 52.53,
    'Estanyeres_C': 142.86,
    'Estanyeres_D': 22.23
}

# Convertir a radianes
rads = np.deg2rad(list(orientaciones.values()))

# Promedio circular
x = np.cos(rads).mean()
y = np.sin(rads).mean()

# Ángulo promedio (en radianes y luego a grados)
mean_angle_rad = np.arctan2(y, x)
mean_angle_deg = np.rad2deg(mean_angle_rad) % 360  # Asegurar valor en [0, 360)

print(f"Orientación promedio circular de Estanyeres: {mean_angle_deg:.2f}°")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from scipy.stats import linregress

# -------------------------------
# ARCHIVOS Y CONFIGURACIÓN
# -------------------------------
archivo_datos = r"M:\material_isa\_validacion\dataset_validation_Molieres_filtrado_diario_indices_v2.csv"
salida_dir = r"M:\material_isa\_validacion\suelo_vs_satelite\regresiones"
os.makedirs(salida_dir, exist_ok=True)

# Archivo donde guardar R2 y p-valores
archivo_resultados = os.path.join(salida_dir, "regresiones_resultados.txt")
with open(archivo_resultados, "w", encoding='utf-8') as f:
    f.write("Resultados de regresiones lineales (R² y p-valor)\n")
    f.write("="*60 + "\n\n")

# -------------------------------
# CARGAR DATOS
# -------------------------------
df = pd.read_csv(archivo_datos, sep=";")
df["date"] = pd.to_datetime(df["date"], dayfirst=True)

# -------------------------------
# COMBINACIONES DE VARIABLES
# -------------------------------
combinaciones = {
    "nivel_vs_indices": ("Nivel agua (mm)", ["NDVI", "NDWI_V", "NDDI"]),
    "humedad_vs_indices": ("SoilMoisture_%", ["NDVI", "NDWI_V", "NDDI"]),
    "suelo_vs_indices": ("SoilTemp_C", ["NDVI", "NDWI_V", "NDDI"])
}

# -------------------------------
# GRAFICAR SUBPLOTS Y REGRESIÓN
# -------------------------------
for nombre_figura, (var_suelo, indices) in combinaciones.items():
    fig, axs = plt.subplots(1, 3, figsize=(10, 4))
    fig.suptitle(f"{var_suelo} vs Índices", fontsize=13, y=0.9)

    for i, indice in enumerate(indices):
        ax = axs[i]
        
        # Filtrar datos válidos
        datos = df[[var_suelo, indice]].dropna()

        # Scatterplot + regresión lineal
        sns.regplot(x=indice, y=var_suelo, data=datos, ax=ax,
                    line_kws={"color": "red"}, scatter_kws={"s": 20, "alpha": 0.5})

        # Quitar etiqueta Y y números de eje Y excepto en el primer subplot
        if i != 0:
            ax.set_ylabel("")
            ax.tick_params(axis='y', labelleft=False)
        else:
            ax.set_ylabel(var_suelo)

        # Título y etiquetas
        #ax.set_title(f"{var_suelo} vs {indice}")
        ax.set_xlabel(indice)
        #ax.set_xlim(-0.2, 1.0)

        # Cálculo R² y p-valor
        slope, intercept, r_value, p_value, std_err = linregress(datos[indice], datos[var_suelo])
        
        # Guardar en TXT
        with open(archivo_resultados, "a", encoding='utf-8') as f:
            f.write(f"{var_suelo} vs {indice}:\n")
            f.write(f"  R² = {r_value**2:.4f}\n")
            f.write(f"  p-valor = {p_value:.4e}\n\n")

        # Mostrar R² y p-valor en el gráfico
        texto_estadistico = f"R² = {r_value**2:.2f}\np = {p_value:.4f}"
        ax.text(0.04, 0.15, texto_estadistico, transform=ax.transAxes,
                fontsize=9, verticalalignment='top',
                bbox=dict(boxstyle="round", facecolor="white", alpha=0.6))

        # Añadir líneas horizontales grises punteadas
        for ytick in ax.get_yticks():
            ax.axhline(y=ytick, color='gray', linestyle='--', linewidth=0.6, alpha=0.5, zorder=0)

    fig.tight_layout(rect=[0, 0, 1, 0.93])
    plt.subplots_adjust(wspace=0.02)
    fig.savefig(os.path.join(salida_dir, f"{nombre_figura}.png"), dpi=300)
    plt.close()

print("✅ Gráficos generados y resultados guardados.")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os

# --- Cargar datos ---
archivo_datos = r"M:\material_isa\_validacion\dataset_validation_Molieres_filtrado_diario_indices_v2.csv"
df = pd.read_csv(archivo_datos, sep=";")
df["date"] = pd.to_datetime(df["date"], dayfirst=True)

# --- Variables a correlacionar ---
indices = ["NDVI", "NDWI_V", "NDDI"]
suelo = ["Nivel agua (mm)", "SoilMoisture_%", "SoilTemp_C"]

# --- Calcular correlación de Spearman ---
cor_matrix = df[suelo + indices].corr(method="spearman")

# Extraer submatriz de correlación: suelo (filas) vs índices (columnas)
cor_sub = cor_matrix.loc[suelo, indices]

# --- Reformatear nombres para visualización (opcional) ---
cor_sub.index = [s.replace("_", " ") for s in cor_sub.index]
cor_sub.columns = [s.replace("_", " ") for s in cor_sub.columns]

# --- Plot heatmap ---
fig, ax = plt.subplots(figsize=(5.2, 3.5))
heatmap = sns.heatmap(
    cor_sub,
    annot=True, fmt=".2f",
    cmap="coolwarm", vmin=-1, vmax=1,
    cbar_kws={"label": "Coef. Spearman", "shrink": 0.7},
    linewidths=0.5, linecolor="gray",
    annot_kws={"fontsize": 8},
    xticklabels=True, yticklabels=True,
    ax=ax
)

# --- Personalizar colorbar ---
cbar = heatmap.collections[0].colorbar
cbar.set_ticks([-1.0, -0.5, 0, 0.5, 1.0])
cbar.ax.tick_params(labelsize=8)
cbar.set_label("Coef. Spearman", fontsize=9)

# --- Etiquetas ---
plt.title("Correlación Spearman: Variables de Suelo vs Índices", fontsize=12)
plt.ylabel("Variable de Suelo", fontsize=10)
plt.xlabel("Índice de Vegetación", fontsize=10)
plt.xticks(fontsize=9)
plt.yticks(fontsize=9)
plt.tight_layout()

# --- Guardar gráfico ---
salida_path = r"M:\material_isa\_validacion\suelo_vs_satelite\heatmap_spearman_suelo_vs_indices.png"
plt.savefig(salida_path, dpi=300)
plt.close()

print(f"✅ Heatmap guardado en: {salida_path}")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import os

# --------------------------------------
# CONFIGURACIÓN
# --------------------------------------
archivo = r"M:\material_isa\_validacion\dataset_validation_Molieres_filtrado_diario_indices_v2.csv"
salida_dir = r"M:\material_isa\_validacion\suelo_vs_satelite\correlacion_desfase"
os.makedirs(salida_dir, exist_ok=True)

# --------------------------------------
# CARGA DE DATOS
# --------------------------------------
df = pd.read_csv(archivo, sep=";")
df["date"] = pd.to_datetime(df["date"], dayfirst=True)
df = df.sort_values("date").reset_index(drop=True)

# Variables
indices = ["NDVI", "NDWI_V", "NDDI"]
variables_suelo = ["SoilMoisture_%", "Nivel agua (mm)", "SoilTemp_C"]
lags = range(-3, 4)  # desde -3 a +3

# --------------------------------------
# CALCULAR CORRELACIONES CON DESFASE
# --------------------------------------
for var in variables_suelo:
    cor_matrix = pd.DataFrame(index=indices, columns=lags)

    for indice in indices:
        for lag in lags:
            # Desplazar el índice (no la variable de suelo)
            shifted = df[indice].shift(lag)
            datos = pd.DataFrame({var: df[var], indice: shifted}).dropna()
            if not datos.empty:
                rho = datos.corr(method="spearman").iloc[0, 1]
                cor_matrix.loc[indice, lag] = rho
            else:
                cor_matrix.loc[indice, lag] = np.nan

    # Convertir a tipo numérico
    cor_matrix = cor_matrix.astype(float)

    # --------------------------------------
    # PLOT HEATMAP
    # --------------------------------------
    fig, ax = plt.subplots(figsize=(6, 3.5))
    sns.heatmap(cor_matrix, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1,
                cbar_kws={"label": "Coef. Spearman"}, ax=ax,
                linewidths=0.5, linecolor="gray")

    ax.set_title(f"Spearman con desfase temporal: {var}", fontsize=11)
    ax.set_xlabel("Desfase (número de pasos)")
    ax.set_ylabel("Índice ND")
    ax.tick_params(labelsize=9)

    plt.tight_layout()
    plt.savefig(os.path.join(salida_dir, f"cor_spearman_desfase_{var.replace(' ', '_')}.png"), dpi=300)
    plt.close()

print("✅ Correlaciones con desfase calculadas y graficadas.")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os

# --- CONFIGURACIÓN ---
archivo = r"M:\material_isa\_validacion\dataset_validation_Molieres_filtrado_diario_indices_v2.csv"
salida_dir = r"M:\material_isa\_validacion\suelo_vs_satelite\correlacion_desfase_dias"
os.makedirs(salida_dir, exist_ok=True)

# --- CARGAR DATOS ---
df = pd.read_csv(archivo, sep=";")
df["date"] = pd.to_datetime(df["date"], dayfirst=True)
df = df.sort_values("date").set_index("date")

# --- VARIABLES ---
variables_suelo = ["Nivel agua (mm)", "SoilMoisture_%", "SoilTemp_C"]
indices = ["NDVI", "NDWI_V", "NDDI"]
desfases_dias = [5, 10, 15, 25, 30, 45, 60, 75, 90]

# --- DESFASES POSITIVOS Y NEGATIVOS ---
desfases_totales = sorted([-d for d in desfases_dias] + [0] + desfases_dias)

# --- FUNCIÓN DE CORRELACIÓN ---
for var_suelo in variables_suelo:
    matriz = pd.DataFrame(index=indices, columns=desfases_totales)

    for indice in indices:
        for dias in desfases_totales:
            # Desfase en días
            indice_shifted = df[indice].shift(freq=pd.Timedelta(days=dias))

            # Combinar y limpiar
            combined = pd.concat([df[var_suelo], indice_shifted], axis=1).dropna()

            if not combined.empty:
                rho = combined.corr(method="spearman").iloc[0, 1]
                matriz.loc[indice, dias] = rho
            else:
                matriz.loc[indice, dias] = float("nan")

    # Convertir a float antes del heatmap
    matriz = matriz.astype(float)

    # --- PLOT ---
    plt.figure(figsize=(10, 4))
    ax = sns.heatmap(
        matriz,
        annot=True, fmt=".2f",
        cmap="coolwarm", center=0,
        cbar_kws={"label": "Coef. Spearman"},
        linewidths=0.5, linecolor="gray"
    )
    plt.title(f"Spearman con desfase (días): {var_suelo}", fontsize=12)
    plt.xlabel("Desfase (días)")
    plt.ylabel("Índice ND")
    plt.tight_layout()

    # --- GUARDAR ---
    nombre_archivo = f"cor_spearman_desfase_dias_{var_suelo.replace(' ', '_').replace('%', '')}.png"
    plt.savefig(os.path.join(salida_dir, nombre_archivo), dpi=300)
    plt.close()

print("✅ Análisis de desfases temporales completado.")
```

```python
#### TEMP suelo y meteo?

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
from scipy.stats import linregress

# -------------------------------
# ARCHIVOS Y CONFIGURACIÓN
# -------------------------------
archivo_datos = r"M:\material_isa\_validacion\dataset_validation_Molieres_filtrado_diario_indices_v2.csv"
salida_dir = r"M:\material_isa\_validacion\suelo_vs_satelite\regresiones"
os.makedirs(salida_dir, exist_ok=True)

# Archivo donde guardar R2 y p-valores
archivo_resultados = os.path.join(salida_dir, "regresiones_temp_suelo_vs_aire.txt")
with open(archivo_resultados, "w", encoding='utf-8') as f:
    f.write("Resultados de regresiones lineales (R² y p-valor)\n")
    f.write("Comparando temperatura del suelo vs temperatura del aire\n")
    f.write("="*70 + "\n\n")

# -------------------------------
# CARGAR DATOS
# -------------------------------
df = pd.read_csv(archivo_datos, sep=";")
df["date"] = pd.to_datetime(df["date"], dayfirst=True)

# -------------------------------
# VARIABLES A ANALIZAR
# -------------------------------
var_suelo = "SoilTemp_C"
var_aire = "MET_Temperature"
indices = ["NDVI", "NDWI_V", "NDDI"]

# -------------------------------
# GRAFICAR SUBPLOTS 2x3
# -------------------------------
fig, axs = plt.subplots(2, 3, figsize=(12, 6))
fig.suptitle("Relación entre índices y temperatura (suelo vs aire)", fontsize=14, y=0.95)

for col, indice in enumerate(indices):
    for row, (var_temp, label) in enumerate([(var_suelo, "Temp. Suelo (°C)"), (var_aire, "Temp. Aire (°C)")]):
        ax = axs[row, col]
        datos = df[[var_temp, indice]].dropna()

        # Scatter + regresión lineal
        sns.regplot(x=indice, y=var_temp, data=datos, ax=ax,
                    line_kws={"color": "red"}, scatter_kws={"s": 20, "alpha": 0.5})

        # Etiquetas eje Y solo en la primera columna
        if col != 0:
            ax.set_ylabel("")
            ax.tick_params(axis='y', labelleft=False)
        else:
            ax.set_ylabel(label)

        # Etiqueta X solo en fila inferior
        if row == 1:
            ax.set_xlabel(indice)
        else:
            ax.set_xlabel("")

        # Cálculo R² y p
        slope, intercept, r_value, p_value, std_err = linregress(datos[indice], datos[var_temp])
        texto_estadistico = f"R² = {r_value**2:.2f}\np = {p_value:.4f}"
        ax.text(0.04, 0.15, texto_estadistico, transform=ax.transAxes,
                fontsize=9, verticalalignment='top',
                bbox=dict(boxstyle="round", facecolor="white", alpha=0.6))

        # Líneas horizontales grises punteadas
        for ytick in ax.get_yticks():
            ax.axhline(y=ytick, color='gray', linestyle='--', linewidth=0.6, alpha=0.5, zorder=0)

        # Guardar resultados
        with open(archivo_resultados, "a", encoding='utf-8') as f:
            f.write(f"{label} vs {indice}:\n")
            f.write(f"  R² = {r_value**2:.4f}\n")
            f.write(f"  p-valor = {p_value:.4e}\n\n")

# -------------------------------
# FORMATO FINAL
# -------------------------------
fig.tight_layout(rect=[0, 0, 1, 0.93])
plt.subplots_adjust(wspace=0.03, hspace=0.15)
fig.savefig(os.path.join(salida_dir, "suelo_vs_aire_vs_indices.png"), dpi=300)
plt.close()

print("✅ Gráfico comparativo generado y resultados guardados.")
```

```python
#------------
# zoom julio-septiembre SERIE TEMPORAL Y REGRESION
#-------------

import pandas as pd
import matplotlib.pyplot as plt
import os
from datetime import datetime
import re
from scipy.stats import linregress
import numpy as np

# -------------------------------
# ARCHIVOS
# -------------------------------
archivo_datos = r"M:\material_isa\_validacion\dataset_validation_Molieres_filtrado_diario_indices_v2.csv"
ruta_nieve = r"M:\material_isa\_validacion\day_in_off_snow_molieres.txt"
salida_dir = r"M:\material_isa\_validacion\suelo_vs_satelite\analisis_julio_septiembre"
os.makedirs(salida_dir, exist_ok=True)

# -------------------------------
# CARGAR DATOS Y FILTRAR FECHAS
# -------------------------------
df = pd.read_csv(archivo_datos, sep=";")
df["date"] = pd.to_datetime(df["date"], dayfirst=True)
fecha_inicio = pd.to_datetime("2024-07-28")
fecha_fin = pd.to_datetime("2024-09-16")
df = df[(df["date"] >= fecha_inicio) & (df["date"] <= fecha_fin)]

# -------------------------------
# CARGAR PERIODOS DE NIEVE
# -------------------------------
def cargar_periodos_nieve(ruta):
    with open(ruta, 'r', encoding='utf-8') as f:
        contenido = f.read()
    periodos = []
    for linea in contenido.splitlines():
        match = re.search(r'Snow in: (\d{2}/\d{2}/\d{4}) \| Snow off: (\d{2}/\d{2}/\d{4})', linea)
        if match:
            snow_in = datetime.strptime(match.group(1), '%d/%m/%Y')
            snow_off = datetime.strptime(match.group(2), '%d/%m/%Y')
            periodos.append((snow_in, snow_off))
    return periodos

periodos_nieve = cargar_periodos_nieve(ruta_nieve)

# -------------------------------
# VARIABLES A GRAFICAR
# -------------------------------
combinaciones = [
    ("Nivel agua (mm)", "nivel"),
    ("SoilMoisture_%", "humedad"),
]

colores_indices = {
    "NDVI": "green",
    "NDWI_V": "blue",
    "NDDI": "gold"
}

# -------------------------------
# GRAFICAR CON REGRESIONES
# -------------------------------
for variable_suelo, etiqueta in combinaciones:
    fig, ax1 = plt.subplots(figsize=(9, 4))
    ax2 = ax1.twinx()

    # --- Suelo ---
    suelo_data = df[["date", variable_suelo]].dropna()
    ax1.scatter(suelo_data["date"], suelo_data[variable_suelo], color="black", label=variable_suelo, s=10, alpha=0.6)
    # Ajustar regresión lineal
    x_num = (suelo_data["date"] - suelo_data["date"].min()).dt.days
    slope, intercept, *_ = linregress(x_num, suelo_data[variable_suelo])
    ax1.plot(suelo_data["date"], intercept + slope * x_num, color="black", linestyle="--", linewidth=1.2)

    ax1.set_ylabel(variable_suelo, color="black")
    ax1.tick_params(axis='y', labelcolor="black")

    # --- Índices ---
    for indice, color in colores_indices.items():
        if indice in df.columns:
            ind_data = df[["date", indice]].dropna()
            ax2.scatter(ind_data["date"], ind_data[indice], label=indice, color=color, s=10, alpha=0.6)
            # Regresión para cada índice
            x_num_i = (ind_data["date"] - ind_data["date"].min()).dt.days
            slope_i, intercept_i, *_ = linregress(x_num_i, ind_data[indice])
            ax2.plot(ind_data["date"], intercept_i + slope_i * x_num_i, color=color, linestyle="--", linewidth=1.2)

    ax2.set_ylabel("Índices ND", fontsize=9)
    ax2.tick_params(axis='y')

    # Título y formato
    ax1.set_title(f"{variable_suelo} vs NDVI / NDWI_V / NDDI (Jul-Sep 2024)", fontsize=11, pad=10)
    ax1.set_xlabel("Fecha")
    ax1.tick_params(axis='x', labelrotation=45, labelsize=8)
    ax1.grid(True, linestyle='--', axis='x', alpha=0.4)

    # Zonas de nieve
    for snow_in, snow_off in periodos_nieve:
        ax1.axvspan(snow_in, snow_off, color='gray', alpha=0.2)
    ax1.text(1.01, 1.02, "Período SNOW", transform=ax1.transAxes,
             fontsize=9, ha='right', color='gray',
             bbox=dict(boxstyle="round,pad=0.2", facecolor='lightgray', edgecolor='none', alpha=0.5))

    # Leyenda
    fig.legend(loc="upper right", bbox_to_anchor=(0.92, 0.41), fontsize=8)

    # Guardar
    fig.tight_layout()
    nombre_archivo = f"{etiqueta}_vs_indices_series_temporales.png"
    fig.savefig(os.path.join(salida_dir, nombre_archivo), dpi=300)
    plt.close()

print("✅ Gráficos con regresiones generados y guardados correctamente.")

```

```python
#### TEMP suelo y meteo? OUTLIER

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os

# --- Cargar datos ---
archivo_datos = r"M:\material_isa\_validacion\dataset_validation_Molieres_filtrado_diario_indices_v2_outlier.csv"
df = pd.read_csv(archivo_datos, sep=";")
df["date"] = pd.to_datetime(df["date"], dayfirst=True)

# --- Variables a correlacionar ---
indices = ["NDVI", "NDWI_V", "NDDI"]
suelo = ["MET_Temperature", "SoilTemp_C"]

# --- Calcular correlación de Spearman ---
cor_matrix = df[suelo + indices].corr(method="spearman")

# Extraer submatriz de correlación: suelo (filas) vs índices (columnas)
cor_sub = cor_matrix.loc[suelo, indices]

# --- Reformatear nombres para visualización (opcional) ---
cor_sub.index = [s.replace("_", " ") for s in cor_sub.index]
cor_sub.columns = [s.replace("_", " ") for s in cor_sub.columns]

# --- Plot heatmap ---
fig, ax = plt.subplots(figsize=(5.2, 3.5))
heatmap = sns.heatmap(
    cor_sub,
    annot=True, fmt=".2f",
    cmap="coolwarm", vmin=-1, vmax=1,
    cbar_kws={"label": "Coef. Spearman", "shrink": 0.7},
    linewidths=0.5, linecolor="gray",
    annot_kws={"fontsize": 8},
    xticklabels=True, yticklabels=True,
    ax=ax
)

# --- Personalizar colorbar ---
cbar = heatmap.collections[0].colorbar
cbar.set_ticks([-1.0, -0.5, 0, 0.5, 1.0])
cbar.ax.tick_params(labelsize=8)
cbar.set_label("Coef. Spearman", fontsize=9)

# --- Etiquetas ---
plt.title("Correlación Spearman: Variables de Suelo vs Índices", fontsize=12)
plt.ylabel("Variable de Suelo", fontsize=10)
plt.xlabel("Índice de Vegetación", fontsize=10)
plt.xticks(fontsize=9)
plt.yticks(fontsize=9)
plt.tight_layout()

# --- Guardar gráfico ---
salida_path = r"M:\material_isa\_validacion\suelo_vs_satelite\heatmap_spearman_suelo_vs_indices_meteo_outlier.png"
plt.savefig(salida_path, dpi=300)
plt.close()

print(f"✅ Heatmap guardado en: {salida_path}")
```

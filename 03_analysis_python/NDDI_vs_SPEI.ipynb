```python
#------------------------
# correlacion Pearson
#-------------------------

import os
import pandas as pd
from scipy.stats import pearsonr
import numpy as np

# -------------------------------
# CONFIGURACIÓN
# -------------------------------
input_dir_nddi = r"M:\material_isa\estadisticos\_Turberas\indices_analisis_NDDI"
input_dir_spei = r"M:\material_isa\estadisticos\anual\sequia\3_12"
output_dir = r"M:\material_isa\estadisticos\_Turberas\SPEI_vs_NDDI\correlacion_pearson"
os.makedirs(output_dir, exist_ok=True)

# -------------------------------
# TURBERAS A PROCESAR
# -------------------------------
turberas = [
    "BassaNera",
    "ClotsdeRialba",
    "Estanyeres",
    "PladeSartari",
    "PlandethHorn",
    "PletadeMolieres"
]

# -------------------------------
# PROCESAMIENTO
# -------------------------------
resultados = []

for turbera in turberas:
    nombre_legible = (turbera
                      .replace("de", "de ")
                      .replace("Pla", "Pla ")
                      .replace("Plan", "Plan ")
                      .replace("Pleta", "Pleta ")
                      .replace("Bassa", "Bassa ")
                      .replace("Clots", "Clots ")
                      .replace("Estanyeres", "Estanyeres")
                      .replace("Rialba", "Rialba")).strip()

    nddi_file = os.path.join(input_dir_nddi, f"NDDI_{turbera}_mensual.csv")
    spei_file = os.path.join(input_dir_spei, f"Sequia_{turbera}_3_12_mensual.csv")

    try:
        df_nddi = pd.read_csv(nddi_file, sep=";")
        df_spei = pd.read_csv(spei_file, sep=";")

        df_nddi["FECHA"] = pd.to_datetime(df_nddi["YEAR_MES"], format="%Y-%m")
        df_spei["FECHA"] = pd.to_datetime(df_spei["fecha"], format="%Y-%m")

        df_merged = pd.merge(
            df_nddi[["FECHA", "NDDI_MEDIA"]],
            df_spei[["FECHA", "spei_3", "spei_12"]],
            on="FECHA", how="inner"
        )

        for spei_col in ["spei_3", "spei_12"]:
            try:
                r, p = pearsonr(df_merged["NDDI_MEDIA"], df_merged[spei_col])
            except Exception:
                r, p = np.nan, np.nan

            resultados.append({
                "Turbera": nombre_legible,
                "X": "NDDI",
                "SPEI": spei_col.upper(),
                "Correlacion": r,
                "P_value": p
            })

    except Exception:
        # Si no se pueden leer los datos, igualmente se agregan filas vacías
        for spei_col in ["spei_3", "spei_12"]:
            resultados.append({
                "Turbera": nombre_legible,
                "X": "NDDI",
                "SPEI": spei_col.upper(),
                "Correlacion": "",
                "P_value": ""
            })

# -------------------------------
# GUARDAR RESULTADO
# -------------------------------
df_out = pd.DataFrame(resultados)
output_file = os.path.join(output_dir, "correlacion_NDDI_vs_SPEI_mensual.csv")
df_out.to_csv(output_file, sep=";", index=False)
print(f"✅ Archivo generado: {output_file}")
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os

# ----------------------
# CONFIGURACIÓN DE RUTAS
# ----------------------
input_csv = r"M:\material_isa\estadisticos\_Turberas\SPEI_vs_NDDI\correlacion_pearson\correlacion_NDDI_vs_SPEI_mensual.csv"
output_dir = r"M:\material_isa\estadisticos\_Turberas\SPEI_vs_NDDI\correlacion_pearson"
os.makedirs(output_dir, exist_ok=True)

# ----------------------
# LECTURA Y FORMATO DE DATOS
# ----------------------
df = pd.read_csv(input_csv, sep=";")

# Renombrar para mayor claridad en el gráfico
renombrar_spei = {
    "SPEI_3": "SPEI 3 meses",
    "SPEI_12": "SPEI 12 meses"
}

# Lista única de turberas
turberas = df["Turbera"].dropna().unique()

# ----------------------
# GENERACIÓN DE HEATMAP POR TURBERA
# ----------------------
for turbera in turberas:
    df_turb = df[df["Turbera"] == turbera]

    # Crear tabla pivote
    matriz = df_turb.pivot(index="X", columns="SPEI", values="Correlacion")
    if matriz.empty:
        continue

    # Renombrar columnas si están en diccionario
    matriz.columns = [renombrar_spei.get(col, col) for col in matriz.columns]

    # Crear heatmap
    plt.figure(figsize=(4, 2.5))
    sns.heatmap(matriz, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1, cbar_kws={"label": "Coef. de Pearson"})

    plt.title(f"Correlación NDDI vs SPEI – {turbera}", fontsize=11)
    plt.xlabel("SPEI")
    plt.ylabel("Índice")
    plt.tight_layout()

    # Guardar gráfico
    nombre_archivo = f"heatmap_pearson_{turbera.lower().replace(' ', '_')}.png"
    output_path = os.path.join(output_dir, nombre_archivo)
    plt.savefig(output_path, dpi=300)
    plt.close()

print("✅ Mapas de calor generados correctamente.")
```

```python
#----------------------
# correlacion anual
#---------------------



import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

# -----------------------
# CONFIGURACIÓN
# -----------------------
nddi_dir = r"M:\material_isa\estadisticos\_Turberas\indices_analisis_NDDI"
spei_dir = r"M:\material_isa\estadisticos\anual\sequia\3_12"
output_dir = r"M:\material_isa\estadisticos\_Turberas\SPEI_vs_NDDI\correlacion_pearson"
os.makedirs(output_dir, exist_ok=True)

# Archivo de salida CSV
output_csv = os.path.join(output_dir, "correlacion_NDDI_vs_SPEI_anual.csv")

# -----------------------
# CALCULAR CORRELACIONES
# -----------------------
registros = []

for file in os.listdir(nddi_dir):
    if file.endswith("_anual.csv") and file.startswith("NDDI_"):
        turbera = file.replace("NDDI_", "").replace("_anual.csv", "")
        turbera_nombre = turbera.replace("Plandeth", "Plan deth").replace("Plade", "Pla de").replace("Pletade", "Pleta de").replace("Clotsde", "Clots de").replace("BassaNera", "Bassa Nera")

        nddi_path = os.path.join(nddi_dir, file)
        spei_path = os.path.join(spei_dir, "Sequia_" + turbera + "_3_12_anual.csv")

        if not os.path.exists(spei_path):
            continue

        df_nddi = pd.read_csv(nddi_path, sep=";")
        df_spei = pd.read_csv(spei_path, sep=";")

        for spei_col in ["spei_3", "spei_12"]:
            if spei_col not in df_spei.columns:
                continue

            merged = pd.merge(df_nddi, df_spei, left_on="YEAR", right_on="YEAR")
            if merged.empty:
                continue

            x = merged["NDDI_MEDIA"]
            y = merged[spei_col]

            if len(x.dropna()) > 2 and len(y.dropna()) > 2:
                corr, pval = pearsonr(x, y)
            else:
                corr, pval = None, None

            registros.append({
                "Turbera": turbera_nombre,
                "X": "NDDI",
                "SPEI": spei_col.upper(),
                "Correlacion": corr,
                "P_value": pval
            })

# Guardar CSV
df_corr = pd.DataFrame(registros)
df_corr.to_csv(output_csv, sep=";", index=False)
print(f"✅ CSV de correlaciones guardado: {output_csv}")

# -----------------------
# GENERAR HEATMAPS
# -----------------------
def plot_heatmap(df, spei_key, titulo, nombre_salida):
    df_spei = df[df["SPEI"] == spei_key]

    pivot = df_spei.pivot(index="Turbera", columns="X", values="Correlacion")

    plt.figure(figsize=(5, len(pivot) * 0.6))
    sns.heatmap(pivot, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1,
                cbar_kws={"label": "Coef. de Pearson"}, linewidths=0.5, linecolor='gray')

    plt.title(titulo, fontsize=12)
    plt.xlabel("Índice")
    plt.ylabel("Turbera")
    plt.tight_layout()

    out_path = os.path.join(output_dir, nombre_salida)
    plt.savefig(out_path, dpi=300)
    plt.close()
    print(f"📊 Guardado: {nombre_salida}")

# Leer CSV nuevamente (por si quieres correr los mapas por separado)
df_corr = pd.read_csv(output_csv, sep=";")

# Mapa de calor para SPEI_3
plot_heatmap(df_corr, "SPEI_3", "Correlación Anual NDDI vs SPEI 3 meses", "heatmap_global_NDDI_vs_SPEI_3_anual.png")

# Mapa de calor para SPEI_12
plot_heatmap(df_corr, "SPEI_12", "Correlación Anual NDDI vs SPEI 12 meses", "heatmap_global_NDDI_vs_SPEI_12_anual.png")
```

```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import spearmanr

# -----------------------
# CONFIGURACIÓN
# -----------------------
nddi_dir = r"M:\material_isa\estadisticos\_Turberas\indices_analisis_NDDI"
spei_dir = r"M:\material_isa\estadisticos\anual\sequia\3_12"
output_dir = r"M:\material_isa\estadisticos\_Turberas\SPEI_vs_NDDI\correlacion_spearman"
os.makedirs(output_dir, exist_ok=True)

# Archivo de salida CSV
output_csv = os.path.join(output_dir, "correlacion_NDDI_vs_SPEI_anual.csv")

# -----------------------
# CALCULAR CORRELACIONES
# -----------------------
registros = []

for file in os.listdir(nddi_dir):
    if file.endswith("_anual.csv") and file.startswith("NDDI_"):
        turbera = file.replace("NDDI_", "").replace("_anual.csv", "")
        turbera_nombre = turbera.replace("Plandeth", "Plan deth").replace("Plade", "Pla de").replace("Pletade", "Pleta de").replace("Clotsde", "Clots de").replace("BassaNera", "Bassa Nera")

        nddi_path = os.path.join(nddi_dir, file)
        spei_path = os.path.join(spei_dir, "Sequia_" + turbera + "_3_12_anual.csv")

        if not os.path.exists(spei_path):
            continue

        df_nddi = pd.read_csv(nddi_path, sep=";")
        df_spei = pd.read_csv(spei_path, sep=";")

        for spei_col in ["spei_3", "spei_12"]:
            if spei_col not in df_spei.columns:
                continue

            merged = pd.merge(df_nddi, df_spei, on="YEAR")
            if merged.empty:
                continue

            x = merged["NDDI_MEDIA"]
            y = merged[spei_col]

            if len(x.dropna()) > 2 and len(y.dropna()) > 2:
                corr, pval = spearmanr(x, y)
            else:
                corr, pval = None, None

            registros.append({
                "Turbera": turbera_nombre,
                "X": "NDDI",
                "SPEI": spei_col.upper(),
                "Correlacion": corr,
                "P_value": pval
            })

# Guardar CSV
df_corr = pd.DataFrame(registros)
df_corr.to_csv(output_csv, sep=";", index=False)
print(f"✅ CSV de correlaciones guardado: {output_csv}")

# -----------------------
# HEATMAPS
# -----------------------
def plot_heatmap(df, spei_key, titulo, nombre_salida):
    df_spei = df[df["SPEI"] == spei_key]
    pivot = df_spei.pivot(index="Turbera", columns="X", values="Correlacion")

    plt.figure(figsize=(5, len(pivot) * 0.6))
    sns.heatmap(pivot, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1,
                cbar_kws={"label": "Coef. de Spearman"}, linewidths=0.5, linecolor='gray')

    plt.title(titulo, fontsize=12)
    plt.xlabel("Índice")
    plt.ylabel("Turbera")
    plt.tight_layout()

    out_path = os.path.join(output_dir, nombre_salida)
    plt.savefig(out_path, dpi=300)
    plt.close()
    print(f"📊 Guardado: {nombre_salida}")

# Leer CSV y generar mapas
df_corr = pd.read_csv(output_csv, sep=";")
plot_heatmap(df_corr, "SPEI_3", "Correlación Anual NDDI vs SPEI 3 meses (Spearman)", "heatmap_global_NDDI_vs_SPEI_3_anual_spearman.png")
plot_heatmap(df_corr, "SPEI_12", "Correlación Anual NDDI vs SPEI 12 meses (Spearman)", "heatmap_global_NDDI_vs_SPEI_12_anual_spearman.png")
```
